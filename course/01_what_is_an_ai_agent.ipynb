## 🤖 Definition

> **An AI agent is a software system in which an LLM autonomously decides — step by step — which tools to call, with which arguments, until a user's goal is reached.**

Key words:

* **autonomously** — there is no hard-coded flow chart. The LLM picks the next step.
* **tools** — Python functions, HTTP APIs, databases, MCP servers, code interpreters, *other agents*…
* **step by step** — the agent runs in a loop, feeding tool results back into the LLM.

An LLM by itself can *think*. An agent can *act*.

## 🔁 The agent loop

```
             ┌──────────────────────┐
  user ────► │  System prompt +     │
             │  conversation +      │
             │  tool descriptions   │
             └────────────┬─────────┘
                          │
                          ▼
                   ┌──────────────┐
                   │     LLM      │  ◄─────── feedback (tool result)
                   └──────┬───────┘                   │
                          │ "I want to call X(args)"  │
                          ▼                           │
                  ┌───────────────┐                   │
                  │  Tool runtime │ ──────────────────┘
                  └───────────────┘
```

Every iteration the LLM sees the **full transcript** (user, assistant, tool messages) and decides:

* call another tool → emit a tool-call message
* respond to the user → emit a final assistant message

The loop stops when the LLM emits a final message (or hits a max-step guardrail).

## 🆚 Agents vs chatbots vs RAG vs workflows

| Pattern        | Who picks the next step?       | When to use                                            |
|----------------|--------------------------------|--------------------------------------------------------|
| **Chatbot**    | Nobody — single LLM call       | Q&A on the model's training data                       |
| **RAG**        | Hard-coded pipeline            | Q&A on YOUR documents; deterministic, cheap            |
| **Workflow**   | You (a DAG)                    | Known, repeatable business processes                   |
| **AI Agent**   | The LLM, every step            | Open-ended tasks that need branching & tool use        |

👉 Rule of thumb: **start with the simplest pattern that works**. Use an agent when you really need autonomy.
