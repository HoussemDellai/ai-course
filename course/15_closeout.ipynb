## 🚀 Suggested next projects

Pick one. Stretch yourself.

1. **Knowledge-base agent**  
   Combine Module 8 (MCP) with an Azure AI Search MCP server. Add memory (Module 11) so the agent remembers user preferences.

2. **Junior-dev agent**  
   Plug the Module 10 shell session into a supervisor that owns a GitHub MCP server. The agent reads issues, writes code, runs tests, opens a PR.

3. **Customer-support orchestrator**  
   Module 13 A2A peers: one agent does product search, one does order lookup, one drafts emails. A supervisor routes the user message.

4. **DevOps agent**  
   Tools to call ARM/Bicep, query Azure Monitor metrics, restart Container Apps. Test against the chaos engineering sandbox first!

## 🧹 Tear down the lab

Don't forget — Foundry, ACA GPU and Cosmos DB cost real money. When you're done experimenting, destroy the lab:

```bash
terraform -chdir=./infra destroy -auto-approve
```

You can always rebuild it with `terraform apply` later.
